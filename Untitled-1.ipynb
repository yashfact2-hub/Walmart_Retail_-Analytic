```python
import	pandas	as	pd
from	sqlalchemy	import	create_engine
#	psql
import	psycopg2
```

```python
df	=	pd.read_csv('Walmart_data.csv',	encoding_errors='ignore')
df.shape
```

```python
df.head()
df.describe()
df.info()
```

```python
#	all	duplicates
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
df.shape
```

```python
df.isnull().sum()
#	dropping	all	rows	with	missing	records
df.dropna(inplace=True)
#	verify
df.isnull().sum()
df.shape
```

```python
df.dtypes
#	unit_price	is	text	with	a	$	sign	->	convert	to	float
df['unit_price']	=	df['unit_price'].str.replace('$',	
'').astype(float)
```

```python
df.head()
```

```python
df.columns
df['total']	=	df['unit_price']	*	df['quantity']
df.head()
```

```python
df.columns
df.columns	=	df.columns.str.lower()
df.columns
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:9554561256p@localhost:5432/walmart"
)

df.to_sql(
    "walmart",
    con=engine,
    if_exists="replace",
    index=False
)
```
